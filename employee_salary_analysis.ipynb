```python
employees=["Asha","Ravi","Meena","Kiran","Swetha","Harsha","Rupesh","Akhil"]
salaries = [45000,52000,38000,75000,60000,42000,58000,90000]
```

```python
total=sum(salaries)
average=total/len(salaries)
highest=max(salaries)
lowest=min(salaries)
top_earner=employees[salaries.index(highest)]
lowest_earner=employees[salaries.index(lowest)]
```

```python
print(f"Total Salary: Rs{total:,}")
print(f"Average Salary: Rs{average:,.2f}")
print(f"Highest Salary: Rs{highest:,}({top_earner})")
print(f"Lowest Salary: Rs{lowest:,}({lowest_earner})")
```

```python
print(f"{'Employee':<10}{'Salary':<10}{'Diff':<10}Status")
for name, sal in zip(employees, salaries):
    diff = sal - average
    if sal > average:
        status = "Above average"
    elif sal < average:
        status = "Below average"
    else:
        status = "Equal to average"
    print(f"{name:<10}{sal:<10}{diff:<+10,.0f}{status}")
```
